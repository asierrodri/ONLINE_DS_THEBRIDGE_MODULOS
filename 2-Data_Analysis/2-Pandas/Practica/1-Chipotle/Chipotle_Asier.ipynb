### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

### Step 3. Name of the max valued product

```python
chipo.item_price = chipo.item_price.str.replace('$', '')
chipo
print(chipo.item_price.dtype)
chipo.item_price = chipo.item_price.astype(float)
print(chipo.item_price.dtype)
```

```python
chipo["precio_unidad"] = chipo.item_price / chipo.quantity
precios_min = chipo.groupby("item_name")["precio_unidad"].min().idxmax()
```

```python
chipo.groupby("item_name")["precio_unidad"].min().idxmax()
```

```python
precios_min = chipo.groupby("item_name")["precio_unidad"].min()
precio_max = precios_min.max()
productos_max = precios_min[precios_min == precio_max]
productos_max
```

### Step 4. How many products cost more than $10.00?

```python
print(len(precios_min[precios_min > 10]))
```
