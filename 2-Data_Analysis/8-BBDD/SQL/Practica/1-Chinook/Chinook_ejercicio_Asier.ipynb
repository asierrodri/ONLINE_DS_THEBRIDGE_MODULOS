```python
import pandas as pd
import sqlite3
```

```python
# Conectamos con la base de datos chinook.db
connection = sqlite3.connect("data/chinook.db")

# Obtenemos un cursor que utilizaremos para hacer las queries
crsr = connection.cursor()
```

```python
# Con esta función leemos los datos y lo pasamos a un DataFrame de Pandas
def sql_query(query):

    # Ejecuta la query
    crsr.execute(query)

    # Almacena los datos de la query 
    ans = crsr.fetchall()

    # Obtenemos los nombres de las columnas de la tabla
    names = [description[0] for description in crsr.description]

    return pd.DataFrame(ans,columns=names)
```

## 1. Facturas de Clientes de Brasil, Nombre del cliente, Id de factura, fecha de la factura y el país de la factura

```python
sql_query('''SELECT firstname, invoiceid, invoicedate, billingcountry
FROM invoices
INNER JOIN customers ON invoices.customerid = customers.customerid
WHERE country = "Brazil"''')
```

## 2. Obtén cada factura asociada a cada agente de ventas con su nombre completo

```python
sql_query('''SELECT i.invoiceid AS FACTURA, e.firstname, e.lastname
          FROM customers as c
          INNER JOIN invoices as i ON c.customerid = i.customerid
          INNER JOIN employees as e ON c.supportrepid = e.employeeid
          ''')
```

## 3. Obtén el nombre del cliente, el país, el nombre del agente y el total

```python
sql_query('''SELECT customers.firstname as NOMBRE_CLIENTE, customers.country, employees.firstname as NOMBRE_AGENTE
          FROM customers
          INNER JOIN employees on supportrepid = employeeid
          ''')
```

## 4. Obtén cada artículo de la factura con el nombre de la canción

```python
sql_query('''SELECT init.invoiceid as FACTURA, t.name
          FROM invoice_items as init
          INNER JOIN tracks as t ON init.trackid = t.trackid
          ''')
```

## 5. Muestra todas las canciones con su nombre, formato, álbum y género

```python
sql_query('''SELECT t.name as NOMBRE_CANCION, m.name as TIPO_ARCHIVO, a.title as ALBUM, g.name as GENERO
          FROM tracks as t
          INNER JOIN media_types as m
          ON t.mediatypeid = m.mediatypeid
          INNER JOIN albums as a
          ON t.albumid = a.albumid
          INNER JOIN genres as g
          ON t.genreid = g.genreid
          ''')
```

## 6. Cuántas canciones hay en cada playlist

```python
sql_query('''SELECT p.name, count(pt.playlistid)
          from playlist_track as pt
          right JOIN playlists as p
          ON pt.playlistid = p.playlistid
          group by p.playlistid
          ''')
```

## 7. Cuánto ha vendido cada empleado

```python
sql_query('''SELECT e.firstname, sum(i.total) as TOTAL_VENTAS
          from customers as c
          inner join invoices as i
          ON c.customerid = i.customerid
          right join employees as e
          ON c.supportrepid = e.employeeid
          group by e.employeeid
          ''')
```
