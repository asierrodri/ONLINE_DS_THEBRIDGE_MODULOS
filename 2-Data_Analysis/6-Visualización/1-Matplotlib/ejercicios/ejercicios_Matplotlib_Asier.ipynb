2. Sabemos que podemos pintar gráficas de dos formas: la figura incluye los ejes o teniendo figura y ejes por separado.

Usando solo una figura, usa numpy para los valores del eje X entre 0 y 5. 

Pinta dos gráficas en dos cajas distintas, a la izquierda una recta con pendiente positiva de 3 que pase por (0,0) y a la derecha una recta con pendiente negativa de 3 que pase por (0,-5). 

Elige la precisión en el eje X que desees.  

```python
# Creamos una única figura
fig = plt.figure(figsize=(10, 4))

# Primer eje (izquierda)
ax1 = plt.subplot(1, 2, 1)
x = np.linspace(0, 5, 100)
y1 = 3 * x   # pendiente +3 y pasa por (0,0)
ax1.plot(x, y1)
ax1.set_title("Recta +3")

# Segundo eje (derecha)
ax2 = plt.subplot(1, 2, 2)
y2 = -3 * x - 5   # pendiente -3 y pasa por (0,-5)
ax2.plot(x, y2)
ax2.set_title("Recta -3")

plt.tight_layout()
plt.show()
```

3. Fija el eje X entre 0 y 5 y el eje Y entre -15 y 15

```python
fig = plt.figure(figsize=(10, 4))

# Subplot izquierda
ax1 = plt.subplot(1, 2, 1)
x = np.linspace(0, 5, 100)
ax1.plot(x, 3*x)
ax1.set_title("Recta +3")

# Límites
ax1.set_xlim(0, 5)
ax1.set_ylim(-15, 15)

# Subplot derecha
ax2 = plt.subplot(1, 2, 2)
ax2.plot(x, -3*x - 5)
ax2.set_title("Recta -3")

# Límites
ax2.set_xlim(0, 5)
ax2.set_ylim(-15, 15)

plt.tight_layout()
plt.show()
```

4. Llama al eje X "eje X", al eje Y "eje Y" y pon de títulos "recta sube" y "recta baja". Muestra dos etiquetas de ejeX pero solo una de eje Y.

```python
fig = plt.figure(figsize=(10, 4))

# Subplot izquierda
ax1 = plt.subplot(1, 2, 1)
x = np.linspace(0, 5, 100)
ax1.plot(x, 3*x)
ax1.set_title("recta sube")

# Etiquetas
ax1.set_xlabel("eje X")
ax1.set_ylabel("eje Y")   # única etiqueta del eje Y

ax1.set_xlim(0, 5)
ax1.set_ylim(-15, 15)

# Subplot derecha
ax2 = plt.subplot(1, 2, 2)
ax2.plot(x, -3*x - 5)
ax2.set_title("recta baja")

# Solo eje X
ax2.set_xlabel("eje X")

ax2.set_xlim(0, 5)
ax2.set_ylim(-15, 15)

plt.tight_layout()
plt.show()
```

11. Prueba a invertir los ejes con ax.invert_axis()
Necesitarás tener los ejes disponibles fuera de la figura

```python
# Creamos figura y eje fuera
fig, ax = plt.subplots(figsize=(5, 3))

amigos = ["Ana", "Luis", "Pedro", "Sofía", "Carmen"]
alturas = [160, 180, 175, 190, 170]

# Barras horizontales
ax.barh(amigos, alturas, color="green")

# Títulos y etiquetas
ax.set_title("Altura de mis amigos")
ax.set_xlabel("Altura (cm)")
ax.set_ylabel("Personas")

# Invertir el eje Y (para que aparezca al revés)
ax.invert_yaxis()

plt.show()
```

13. Vamos a introducir pandas. Crea un DataFrame con las columnas "Year" de valores 2015, 2016, 2017, 2018, 2019 y la columna "Sold_items_A" de valores 1000, 3500, 4000, 5500, 7000

```python
data = {
    "Year": [2015, 2016, 2017, 2018, 2019],
    "Sold_items_A": [1000, 3500, 4000, 5500, 7000]
}

df = pd.DataFrame(data)
df
```

14. Ahora pintamos un gráfico de línea con las ventas respecto al año. Pon un título y etiquetas en los ejes.
Cambia el estilo a 'seaborn-white'. La línea debe ser con rayas y verde.

```python
# Crear figura
plt.figure(figsize=(6, 4))

# Gráfico de línea
plt.plot(df["Year"], df["Sold_items_A"], linestyle="--", color="green")

# Título + ejes
plt.title("Ventas por Año")
plt.xlabel("Año")
plt.ylabel("Unidades Vendidas")

plt.show()
```

15. Otro departamento B ha vendido en esos años 2000, 3100, 5000, 4000, 6000 unidades. Incluye esa columna en el DataFrame y pinta en la misma gráfica las dos líneas. B es una línea punteada y roja. Muestra una leyenda abajo a la derecha.

```python
# Añadir al df existente
df["Sold_items_B"] = [2000, 3100, 5000, 4000, 6000]

# Figura
plt.figure(figsize=(6, 4))

# Línea A
plt.plot(df["Year"], df["Sold_items_A"], linestyle="--", color="green", label="A")

# Línea B
plt.plot(df["Year"], df["Sold_items_B"], linestyle=":", color="red", label="B")

# Título + etiquetas
plt.title("Ventas por Año")
plt.xlabel("Año")
plt.ylabel("Unidades Vendidas")

# Leyenda abajo derecha
plt.legend(loc="lower right")

plt.show()
```

19. Vamos a pintar un histograma de unas alturas generadas aleatoriamente. Crea 300 alturas entre 150 y 190 cm.

Pista: https://pandas.pydata.org/pandas-docs/version/0.23.4/generated/pandas.DataFrame.plot.html

```python
# Generar 300 alturas entre 150 y 190 cm
alturas = np.random.randint(150, 191, 300)

# Crear DataFrame
df = pd.DataFrame({"Altura": alturas})

# Histograma
df["Altura"].plot(kind="hist", bins=15)

plt.title("Histograma de alturas")
plt.xlabel("Altura (cm)")
plt.ylabel("Frecuencia")
plt.show()
```
