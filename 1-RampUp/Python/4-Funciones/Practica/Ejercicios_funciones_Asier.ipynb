## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
nombres_dias = ['lunes', 'martes', 'miercoles', 'jueves', 'viernes', 'sabado', 'domingo']

def convertir_semana(numero):

    '''
    Función que convierte los números entre 0 y 6 en sus respectivos días de la semana
    Inputs:
        numero: int
        
    Output:
        dia: str
    '''

    if numero > 6 or numero < 0:
        print("Introduce un número entre 0 y 6")
    else:
        dia = nombres_dias[numero]
        return dia

print(convertir_semana(int(input("Introduce un número entre 0 y 6"))))
print(convertir_semana.__doc__)
````

## Ejercicio 2
En el ejercicio 8 de flujos de control, creábamos una pirámide invertida de números desde el 5. Crea una función que replique el comportamiento de la pirámide, y utiliza un único parámetro de entrada de la función para determinar el número de filas de la pirámide.

````python
def piramide(filas):

    '''
    Función que crea una piramide de números inversos desde el número que tu introduzcas.
    En cada fila se quita el primer número
    Inputs:
        filas: int
        
    Output:
        print: int
        print: salto de linea
    '''

    lista = list(range(filas,0,-1))
    for i in lista:
        for j in (range(i,0,-1)):
            print(j, end=" ")
        print()

filas = int(input("De cuántas filas quieres la piramide?"))
piramide(filas)
print(piramide.__doc__)
````

## Ejercicio 3
Escribe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero sea  mayor que el segundo, o que el segundo sea mayor que el primero

````python
def comparar_numeros(numero_1, numero_2):
    if numero_1 == numero_2:
        print("Los números son iguales")
    elif numero_1 > numero_2:
        print("El primer número es mayor que el segundo")
    else:
        print("El segundo número es mayor que el primero")

comparar_numeros(int(input("Introduce un número")), int(input("Introduce otro número")))
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea una letra que contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayúscula, como minúscula

````python
def contador_letras(texto, letra):
    i = 0
    for l in texto:
        if l == letra:
            i +=1
    return i

texto = input("Introduce un texto")
letra = (input("Introduce la letra a buscar"))

contador_letras(texto, letra)
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string, siendo la clave la letra y el valor el conteo.

````python
def crear_diccionario(frase):
    diccionario = {}

    for i, letra in enumerate(frase):
        diccionario[letra] = contador_letras(frase, letra)
    return diccionario
    
frase = input("Introduce una frase")
crear_diccionario(frase)
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

Tendrá como salida la lista

````python
def modificar_lista(lista, comando, elemento):
    if comando == "add":
        lista.append(elemento)
    elif comando == "remove":
        lista.remove(elemento)
    else:
        return "Comando incorrecto"
    
    return lista

lista = [1, 2, True, "caballo"]
elemento = input("Introduce elemento a añadir o eliminar")
comando = input("Introduce add o remove")

modificar_lista(lista, comando, elemento)
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

fibonacci(n) = fibonacci(n-1) + fibonacci(n-2)

````python
def fibonacci(n):
    if n == 0:
        return n
    elif n == 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

fibonacci(4)
````

## Ejercicio 9
Define en una única celda las siguientes funciones:
* Función que calcule el área de un cuadrado
* Función que calcule el área de un triángulo
* Función que calcule el área de un círculo

En otra celda, calcular el area de:
* Dos círculos de radio 10 + un triángulo de base 3 y altura 7
* Un cuadrado de lado = 10 + 3 círculos (uno de radio = 4 y los otros dos de radio = 6) + 5 triángulos de base = 2 + altura = 4

````python
import math

def cuadrado(lado):
    return lado**2

def triangulo(base, altura):
    return 1/2 * base * altura

def circulo(radio):
    return math.pi * (radio**2)
````

````python
area_circulo = circulo(10)*2
area_triangulo = triangulo(3, 7)
print(area_circulo + area_triangulo)

area_cuadrado = cuadrado(10)
area_circulos = circulo(4) + 2*(circulo(6))
area_triangulo = triangulo(2, 4)*5

area_total = area_cuadrado + area_circulos + area_triangulo
print(area_total)
````
