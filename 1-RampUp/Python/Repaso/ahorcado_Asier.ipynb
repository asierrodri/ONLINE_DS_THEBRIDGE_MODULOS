## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
def pintar_ahorcado(lista_letras, letras_usadas):
    for l in lista_letras:
        print(l, end=" ")
    print("Letras usadas: ", end=" ")
    for l in letras_usadas:
        print(l, end=" ")
    print()

def letra_en_palabra(palabra, lista_letras, letras_usadas, letra):
    posiciones = []
    for i, l in enumerate(palabra):
        if l == letra:
            posiciones.append(i)
        contador = 0
        while contador < len(posiciones):
            for l in lista_letras:
                lista_letras[posiciones[contador]] = letra
            contador += 1
    pintar_ahorcado(lista_letras, letras_usadas)
            

def jugar(palabra, lista_letras, letras_usadas, vidas):
    print("Adivina la palabra:")
    pintar_ahorcado(lista_letras, letras_usadas)

    while vidas > 0:
        print()
        letra = input("Introduce una letra").lower()
        if len(letra) != 1:
            print("Solo una letra")

        elif letra in lista_letras or letra in letras_usadas:
            print("Ya habías introducido esa letra")

        elif letra in palabra:
            letra_en_palabra(palabra, lista_letras, letras_usadas, letra)
            if "_" not in lista_letras:
                print("HAS GANADO")
                break
            
        elif letra not in palabra:
            print("Has perdido una vida")
            vidas -= 1
            print(f"Te quedan {vidas} vidas")
            letras_usadas.append(letra)
            if vidas > 0:
                pintar_ahorcado(lista_letras, letras_usadas)
            else:
                print("HAS PERDIDO")
        else:
            print("Error desconocido")
````

````python
palabra = input("Introduce la palabra").lower()
lista_letras = list("_" * len(palabra))
letras_usadas = []
vidas = 6

jugar(palabra, lista_letras, letras_usadas, vidas)
````
