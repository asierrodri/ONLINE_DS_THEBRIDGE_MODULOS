```python
alfabeto = ["A", "B", "C", "D", "E", "F", "G", "H", "I","J"]

i = int(input("Introduce la primera coordenada"))
j = int(input("Introduce la segunda coordenada"))

if 0 > i > 9 or j > 9:
    print("Coordenadas incorrectas")
elif tablero[i][j] == "B":
    tablero[i][j] = "x"
    print("Tocado en posición " + str(i) + "," + str(j))
elif tablero[i][j] == " ":
    tablero[i][j] = "o"
    print("Agua")
else:
    print("Ya había introducido esas coordenadas")

i = 1
print("*", end = " ")
while i<11:
    print(i, end=" ")
    i += 1
print()
x = 0
for p in tablero:
    print(alfabeto[x], end=" ")
    for i in p:
        print(i, end=" ")
    print("|")
    x += 1
print("- "*12)
```
