```python
def tablero(filas):
    alfabeto = ["A", "B", "C", "D", "E", "F", "G", "H", "I","J"]
    x = 1
    print(" ", end = " ")
    while x < filas + 1:
        print(x, end = " ")
        x += 1
    print()
    x = 0
    while x < filas:
        i = 0
        print(alfabeto[x], end=" ")
        while i < filas:
            print("O", end = " ")
            i +=1
        print()
        x += 1

tablero(10)
```
